````python
print("(c) Cho hai đỉnh A, B. Hỏi hai đỉnh A, B có kề nhau không?")

def LaKe(G, a, b):
    """
    input: G, a, b
    return:
    + -1: dinh a, hay b khong ton tai
    +  0: dinh a, b ton tai va a khong ke b
    +  1: dinh a, b ton tai va a, b ke nhau
    """
    result = None
    """ CÁC BẠN LÀM BÀI Ở ĐÂY """
    # ...
    # Kiểm tra tồn tại
    if a not in G or b not in G:
        return -1

    # Kiểm tra kề nhau
    if b in G[a]:
        return 1
    else:
        return 0

    return result
````

````python
print("1. Lưu đồ thị xuống tập tin")

def LuuDoThi(G, file_path, verbose = True):
    import os

    # Tạo thư mục chứa file_path nếu cần
    file_dir = os.path.dirname(file_path)
    if file_dir != "" and not os.path.exists(file_dir):
        os.makedirs(file_dir)
        if verbose:
            print(f"+ Tạo thư mục: {file_dir}")

    # Lưu đồ thị
    with open(file_path, "wt") as file:
        # Ghi số lượng đỉnh
        file.write(str(len(G)) + "\n")

        # Ghi từng đỉnh và danh sách kề
        for u, neighbors in G.items():
            line = " ".join([u] + neighbors)  # nối u và các neighbors thành 1 dòng
            file.write(line + "\n")

        if verbose:
            print(f"Lưu đồ thị thành công xuống tập tin: {file_path}")
    pass

# --- Test ---
LuuDoThi(ADJ, "dske1.txt", verbose = True)
with open("dske1.txt", "rt") as file:
    lines = file.readlines()
    for line in lines:
        print(line, end="")
````

````python
print("2. Đọc đồ thị từ tập tin")
import pprint

def DocDoThi(file_path, verbose = True):
    """
    return:
    + None: đọc thất bại
    + <>None: trả về đồ thị dạng dict
    """
    import os

    result = None
    if not os.path.exists(file_path):
        result = None
    else:
        G = {}
        with open(file_path, "rt") as file:
            n = int(file.readline().strip())  # đọc số lượng đỉnh
            for _ in range(n):
                line = file.readline().strip()
                if line == "":
                    continue
                parts = line.split()
                vertex = parts[0]
                neighbors = parts[1:] if len(parts) > 1 else []
                G[vertex] = neighbors  # lưu vào dict

        if verbose:
            print(f"Đọc đồ thị thành công từ: {file_path}")
        result = G
    return result

# --- Test ---
G1 = DocDoThi("dske1.txt", verbose = True)
pprint.pprint(G1)
````

### Bài 2. Tìm kiếm đường đi trên đồ thị

#### 1. Tìm kiếm theo chiều rộng (BFS)

````python
import pprint

def BFS(G, start, goal):
    """
    return:
    + path[v] = u nghĩa là để đến v phải qua u
    + {} nếu không tìm thấy đường đi
    + None nếu start hoặc goal không hợp lệ
    """
    if G.get(start) is None or G.get(goal) is None:
        return None  # đỉnh không hợp lệ

    path = {}      # path[v] = u
    s_open = []    # hàng đợi BFS
    s_closed = []  # danh sách đã duyệt

    # đưa start vào open
    s_open.append(start)
    path[start] = None

    while len(s_open) > 0:
        u = s_open.pop(0)    # lấy đỉnh đầu tiên từ hàng đợi
        s_closed.append(u)

        # duyệt tất cả các đỉnh kề với u
        for v in G[u]:
            if v not in s_open and v not in s_closed:
                s_open.append(v)   # thêm v vào hàng đợi để duyệt tiếp
                path[v] = u        # v đến từ u

    return path




````

````python
path = BFS(ADJ, "A", "B")
pprint.pprint(path)
````

````python
def find_path(path, start, goal):
    """
    Cho mang truy hoi duong, tim ds dinh tu start --> goal
    """
    result = []

    """ CÁC BẠN LÀM BÀI Ở ĐÂY """
    # ...
    # Nếu start hoặc goal không có trong path -> không tìm được đường
    if start not in path or goal not in path:
        return result

    # Bắt đầu từ goal lần ngược về start
    current = goal
    while current is not None:
        result.insert(0, current)   # chèn vào đầu danh sách
        current = path[current]     # đi ngược về đỉnh trước đó

    # Kiểm tra xem điểm bắt đầu có khớp với start không
    if result[0] != start:
        return []  # không có đường từ start đến goal

    return result
    pass
````

#### 2. Tìm kiếm theo chiều rộng (DFS)

````python
import pprint

def DFS(G, start, goal):
    """
    return:
    + mang chua duong di tu a --> b, neu rong la ko co duong di
    + None: dinh a, b khong hop le
    """
    result = None

    if G.get(start) is None or G.get(goal) is None:
        result = None
    else:
        path = {}     # path[a]=b nghia la muon di toi dinh b phai qua a
        s_open   = []
        s_closed = []

        # đưa start vào open
        s_open.append(start)
        path[start] = None
        while len(s_open)>0:
            """ CÁC BẠN LÀM BÀI Ở ĐÂY """
            # ...
            u = s_open.pop()      # lấy phần tử cuối cùng trong stack (DFS)
            s_closed.append(u)

            # duyệt các đỉnh kề với u
            for v in G[u]:
              if v not in s_open and v not in s_closed:
                  s_open.append(v)  # thêm vào stack
                  path[v] = u       # muốn đến v phải qua u


    return path
````
